# GQ-Dreamer · Pokémon Red — play the trained model

Loads the trained agent from `model/model.pt` and lets it play in an emulator window. The model file is stored with Git LFS: after cloning, run `git lfs pull` once. A full episode is long (10,000+ agent steps); `MAX_STEPS` stops it earlier. Set `RECORDING = True` to also save an MP4 to `recordings/`.

```python
from gq_dreamer import run

MODEL = "model/model.pt"
MAX_STEPS = 3000      # agent steps to play (None: the whole episode)
RECORDING = False     # save recordings/eval_agent1.mp4
WINDOW = True         # False: headless (e.g. on a server, together with RECORDING = True)
VERBOSE = True        # print maps entered, events and badges

trainer = run(mode="eval", checkpoint=MODEL, max_steps=MAX_STEPS, recording=RECORDING, window=WINDOW,
              verbose=VERBOSE)
```
